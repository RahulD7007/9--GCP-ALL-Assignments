# GCS Operations — Google cloud SDK Shell:- 

How to perform the following GCS operations using the **`gcloud` command-line tool** :

1. Create a bucket
2. List buckets
3. Upload an object
4. Delete an object
5. Copy an object
6. Move an object

## Steps:-
- `gcloud` CLI installed.
- Authenticated: `gcloud auth login`
- Correct project set: `gcloud config set project YOUR_PROJECT_ID`

> gcloud storage command reference: https://cloud.google.com/sdk/gcloud/reference/storage

## bash
#### Verify gcloud is installed and check the active configuration
> gcloud --version

> gcloud config list

## 1. Create a Bucket

````python
gcloud storage buckets create gs://gcp-aiml-dummydatagcsrahul06082026 --default-storage-class=NEARLINE --location=US-CENTRAL1 --enable-hierarchical-namespace --uniform-bucket-level-access --public-access-prevention
````

## 3. Upload an Object

```python
gcloud storage cp newfile.txt gs://gcp-aiml-dummydatagcs1/
```






#### Update an Object:- 

```python
gcloud storage buckets update gs://gcp-aiml-dummydatagcs1
```





````python
gcloud storage cp newfile.txt gs://gcp-aiml-dummydatagcsrahul2026/* gs://destination-bucketrahuld2026
````

---
## 4. Copy an Object:- 

### Create a new bucket:-
```python
gcloud storage buckets create gs://destination-bucket
```

### Then copy objects:- 
```python
gcloud storage cp newfile.txt gs://gcp-aiml-dummydatagcs1/* gs://destination-bucket
```

## 5. Move an Object:- 

```python
gcloud storage mv gs://gcp-aiml-dummydatagcs1/newfile.txt gs://gcp-aiml-dummydatagcs/
```

## ALL Commands:-

| Operation      | gcloud command                                       |
|----------------|-------------------------------------------------------|
| Create bucket  | `gcloud storage buckets create gs://gcp-aiml-dummydatagcs1 --default-storage-class=NEARLINE --location=US-CENTRAL1 --enable-hierarchical-namespace --uniform-bucket-level-access --public-access-prevention`            |
| List buckets   | `gcloud storage buckets list`                          |
| Upload object  | `gcloud storage cp newfile.txt gs://gcp-aiml-dummydatagcs1/`      |
| Delete object  | `gcloud storage rm gs://destination-bucket/newfile.txt`                 |
| Copy object    | `gcloud storage cp --recursive gs://gcp-aiml-dummydatagcs1/* gs://destination-bucket`                  |
| Move object    | `gcloud storage mv gs://gcp-aiml-dummydatagcs1/newfile.txt gs://gcp-aiml-dummydatagcs/`                  |
