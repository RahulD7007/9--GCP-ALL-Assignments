# Google Cloud Storage (GCS) Operations — UI (Google Cloud Console)

This notebook documents how to perform the following GCS operations using the **Google Cloud Console (Web UI)**:

1. Create a bucket
2. List buckets
3. Upload an object
4. Delete an object
5. Copy an object
6. Move an object

Console URL: https://console.cloud.google.com/storage/browser


---
## 1. Create a Bucket

1. Go to the [Cloud Storage Buckets page]
2. Click **+ CREATE** (Create bucket).
3. Enter a globally unique **bucket name** (e.g. `gcp-aiml-dummydatagcs`).
4. Choose where to store data (**Location type** — Region and the **Location**, e.g. `us-central1`).
5. Choose a **storage class** (Standard).
6. Choose **access control** (Uniform).
7. Click **CREATE**.



---
## 2. List Buckets

1. Go to the [Cloud Storage Buckets page](https://console.cloud.google.com/storage/browser).
2. Ensure the correct project is selected.
3. All buckets belonging to the project are listed in the table, along with their location, storage class, and creation date.
4. Use the search/filter box at the top of the table to filter by bucket name if you have many buckets.



---
## 3. Upload an Object

1. Go to the [Cloud Storage Buckets page](https://console.cloud.google.com/storage/browser) and click the target bucket.
2. Click **UPLOAD** > **Upload files** (or **Upload folder** for multiple files/folders).
3. Select the file(s) from your local machine in the file picker.
4. The upload progress is shown at the bottom of the page; once complete, the object(s) appear in the bucket listing.
5. Optionally, drag and drop files directly onto the bucket's object list area to upload.





---
## 4. Delete an Object

1. Navigate to the bucket containing the object.
2. Select the checkbox next to the object (or objects) you want to delete.
3. Click **DELETE** in the toolbar above the object list.
4. Confirm the deletion in the dialog that appears.





---
## 5. Copy an Object

1. Navigate to the bucket containing the object.
2. Select the checkbox next to the object you want to copy.
3. Click the **⋮** (more actions) menu, or the **COPY** button in the toolbar.
4. In the **Copy object** dialog, choose the destination bucket and/or folder path, and optionally rename the object.
5. Click **COPY**. The original object remains in place, and a new copy is created at the destination.



---
## 6. Move an Object

> The Cloud Console UI does not have a single dedicated "Move" action across buckets; a cross-bucket move is performed as a **copy followed by a delete** of the original. Within the same bucket, **Rename** effectively moves an object to a new path/folder.

1. Navigate to the bucket containing the object.
2. Select the checkbox next to the object you want to move.
3. Click the **⋮** (more actions) menu and choose **Rename** (to move/rename within the same bucket, including moving into a different "folder" path).
4. If moving to a **different bucket**:
   - Use **Copy** to copy the object to the destination bucket (see section 5 above).
   - Then go back to the source object, select it, and click **DELETE** to remove it from the source bucket.
