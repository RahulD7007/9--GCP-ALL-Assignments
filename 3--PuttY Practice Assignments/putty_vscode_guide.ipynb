# 0. Golden Rules — Remember These

| Area | Golden rule |
|---|---|
| GCP | `gcloud` manages Google Cloud resources |
| SSH | `ssh` gives you a remote terminal |

# VSCode Remote-SSH setup

```bash
# From VSCode's integrated terminal, generate SSH config entries for ALL your GCE instances
gcloud compute config-ssh

# This writes entries into ~/.ssh/config that look like:
#   Host instance-20260816-051731.us-central1-c.gcp-aiml-sandbox-rahuld-1234
#     HostName 34.27.152.57
#     IdentityFile ~/.ssh/google_compute_engine
#     UserKnownHostsFile=~/.ssh/google_compute_known_hosts
#     User ml_dev_user

# Connect directly from a terminal using the generated host alias
ssh instance-20260816-051731.us-central1-c.gcp-aiml-sandbox-rahuld-1234

# Then in VSCode:
#   Ctrl/Cmd+Shift+P -> "Remote-SSH: Connect to Host..." -> pick the same alias
```

# SSH File Transfer — SCP
```bash
# Local → Remote
scp local_file.txt USERNAME@EXTERNAL_IP:/home/USERNAME/

# Remote → Local
scp USERNAME@EXTERNAL_IP:/home/USERNAME/file.txt .

# Local directory → Remote
scp -r myfolder USERNAME@EXTERNAL_IP:/home/USERNAME/

# Specific private key
scp -i ~/.ssh/YOUR_PRIVATE_KEY local_file.txt \
    USERNAME@EXTERNAL_IP:/home/USERNAME/

# Custom SSH port
scp -P 22 local_file.txt USERNAME@EXTERNAL_IP:/home/USERNAME/
```

# GCP Copying Files with gcloud

```bash
# Local → VM
gcloud compute scp local_file.txt VM_NAME:/home/USERNAME/ \
  --zone=ZONE

# VM → local
gcloud compute scp VM_NAME:/home/USERNAME/file.txt . \
  --zone=ZONE

# Directory
gcloud compute scp --recurse ./myfolder VM_NAME:/home/USERNAME/ \
  --zone=ZONE
```

# GCP Storage — Basic gcloud storage

```bash
# List buckets
gcloud storage buckets list

# List objects
gcloud storage ls gs://BUCKET_NAME/

# Upload
gcloud storage cp local_file.txt gs://BUCKET_NAME/

# Upload directory recursively
gcloud storage cp --recursive ./myfolder gs://BUCKET_NAME/myfolder/

# Download
gcloud storage cp gs://BUCKET_NAME/file.txt .

# Download directory
gcloud storage cp --recursive gs://BUCKET_NAME/myfolder/ ./myfolder/

# Delete object
gcloud storage rm gs://BUCKET_NAME/file.txt

```

# 65. Interview / Viva Questions

### Q1. What is PuTTY?
A GUI SSH client commonly used on Windows to connect to remote servers.

### Q2. What is PuTTYgen?
A utility used to generate/manage SSH key pairs.

### Q3. Public key vs private key?
Public key can be installed on the server; private key must remain secret.

### Q4. What does `ssh` do?
It establishes a secure remote shell connection.

### Q5. What does `scp` do?
It securely copies files over SSH.

### Q6. What does `gcloud compute ssh` do?
It provides a Google Cloud-oriented way to SSH into a Compute Engine VM.

### Q7. What does `gcloud compute config-ssh` do?
It configures SSH entries for GCE instances so they can be accessed through generated host aliases.

### Q8. What does VS Code Remote-SSH provide?
It lets VS Code work with files and terminals on a remote machine.

### Q9. What does `chmod +x` do?
Adds execute permission to a file.

### Q10. Why use a virtual environment?
To isolate Python dependencies for a project.
